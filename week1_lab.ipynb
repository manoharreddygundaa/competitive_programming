```python
data = """1
8 3
101 101 102 102 102 105 110 110
101
102
999
"""

tokens = data.split()
ptr = 0

def nextint():
    global ptr
    val = int(tokens[ptr])
    ptr += 1
    return val


def first_occurrence(arr, x):
    low, high = 0, len(arr) - 1
    ans = -1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == x:
            ans = mid
            high = mid - 1
        elif arr[mid] < x:
            low = mid + 1
        else:
            high = mid - 1
    return ans


def last_occurrence(arr, x):
    low, high = 0, len(arr) - 1
    ans = -1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == x:
            ans = mid
            low = mid + 1
        elif arr[mid] < x:
            low = mid + 1
        else:
            high = mid - 1
    return ans


T = nextint()

for _ in range(T):
    N = nextint()
    Q = nextint()

    arr = [nextint() for _ in range(N)]

    for _ in range(Q):
        X = nextint()

        first = first_occurrence(arr, X)
        if first == -1:
            print(0)
        else:
            last = last_occurrence(arr, X)
            print(last - first + 1)
```

```python
import sys
sys.setrecursionlimit(10**7)

# ---------- INPUT READER ----------
INPUT = """1
9
-2 1 -3 4 -1 2 1 -5 4
"""

data = INPUT.split()
ptr = 0

def nextint():
    global ptr
    val = int(data[ptr])
    ptr += 1
    return val


# ---------- DIVIDE & CONQUER ----------

def max_crossing_sum(arr, left, mid, right):
    s = 0
    left_max = -10**18
    for i in range(mid, left - 1, -1):
        s += arr[i]
        left_max = max(left_max, s)

    s = 0
    right_max = -10**18
    for i in range(mid + 1, right + 1):
        s += arr[i]
        right_max = max(right_max, s)

    return left_max + right_max


def max_subarray(arr, left, right):
    if left == right:
        return arr[left]

    mid = (left + right) // 2

    left_ans = max_subarray(arr, left, mid)
    right_ans = max_subarray(arr, mid + 1, right)
    cross_ans = max_crossing_sum(arr, left, mid, right)

    return max(left_ans, right_ans, cross_ans)


# ---------- DRIVER ----------
T = nextint()

for _ in range(T):
    N = nextint()
    arr = [nextint() for _ in range(N)]

    print(max_subarray(arr, 0, N - 1))
```
