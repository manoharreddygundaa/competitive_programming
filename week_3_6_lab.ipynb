```python
import sys
from io import StringIO

def main():
    input = sys.stdin.readline

    # -------- Segment Tree --------
    def build(arr):
        n = len(arr)
        size = 1
        while size < n:
            size *= 2

        seg = [0] * (2 * size)

        for i in range(n):
            seg[size + i] = arr[i]

        for i in range(size - 1, 0, -1):
            seg[i] = seg[2*i] + seg[2*i + 1]

        return seg, size


    def update(seg, size, idx, val):
        pos = size + idx
        seg[pos] = val

        pos //= 2
        while pos:
            seg[pos] = seg[2*pos] + seg[2*pos + 1]
            pos //= 2


    def query(seg, size, l, r):
        l += size
        r += size
        ans = 0

        while l <= r:
            if l % 2 == 1:
                ans += seg[l]
                l += 1
            if r % 2 == 0:
                ans += seg[r]
                r -= 1
            l //= 2
            r //= 2

        return ans

    # -------- MAIN --------
    T = int(input())
    output = []

    for _ in range(T):

        N = int(input())
        arr = list(map(int, input().split()))

        seg, size = build(arr)

        Q = int(input())

        for _ in range(Q):
            q = list(map(int, input().split()))

            if q[0] == 1:
                _, i, x = q
                update(seg, size, i, x)   # 0-based

            else:
                _, l, r = q
                output.append(str(query(seg, size, l, r)))

    print("\n".join(output))


# ---------- FAKE INPUT ----------
input_data = """1
5
1 3 5 7 9
4
2 1 3
1 2 10
2 1 3
2 0 4
"""

sys.stdin = StringIO(input_data)
main()
```
